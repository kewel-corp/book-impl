## 2.2　ファイルとディレクトリを操作する

```bash
pwd              # 現在いるディレクトリを表示（print working directory）
ls               # 現在のディレクトリの中身を一覧表示（list）
ls -l            # 詳細情報付きで一覧表示
ls -la           # 隠しファイルも含めて詳細表示
cd /home/alice # 指定したディレクトリに移動（change directory）
cd ..            # 一つ上のディレクトリに移動
cd ~             # ホームディレクトリに移動
```

```bash
mkdir dataset            # dataset というディレクトリを作る
mkdir -p data/train/img  # 途中のディレクトリごと一気に作る
cp a.txt b.txt           # a.txt を b.txt という名前でコピー
cp -r dir1 dir2          # ディレクトリ dir1 を丸ごとコピー
mv old.txt new.txt       # 名前を変更（または移動）
rm file.txt              # ファイルを削除
rm -r dir                # ディレクトリを丸ごと削除
```

## 2.3　ファイルの中身を見る

```bash
cat log.txt        # ファイルの中身をすべて表示
head -20 data.csv  # 先頭20行だけ表示
tail -20 data.csv  # 末尾20行だけ表示
tail -f train.log  # 追記されるログをリアルタイムに追う（学習の進捗確認に必須）
less big.txt       # 大きなファイルをスクロールしながら閲覧（q で終了）
wc -l data.csv     # 行数を数える（症例数の確認などに便利）
```

## 2.6　プロセスとGPUの状態を見る

```bash
top              # CPUやメモリの使用状況をリアルタイム表示
htop             # top の見やすい版（要インストール）
nvidia-smi       # GPUの使用状況・VRAM・温度を表示
watch -n 1 nvidia-smi  # 1秒ごとにGPU状態を更新表示
ps aux | grep python   # 実行中のPythonプロセスを探す
kill 12345       # プロセス番号12345を終了させる
```

## 長時間の学習を安全に回す ― screen と nohup

```bash
# screen を使う場合
screen -S train        # "train" という名前の仮想端末を作る
python train.py        # この中で学習を開始
# Ctrl+A → D で切り離す（学習は継続）。screen -r train で復帰

# nohup を使う場合
nohup python train.py > train.log 2>&1 &   # 閉じても継続、ログを記録
tail -f train.log                          # 進捗をリアルタイムに追う
```
