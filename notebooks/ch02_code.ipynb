## 2.2　ファイルとディレクトリを操作する

```bashpwd              # 現在いるディレクトリを表示（print working directory）ls               # 現在のディレクトリの中身を一覧表示（list）ls -l            # 詳細情報付きで一覧表示ls -la           # 隠しファイルも含めて詳細表示cd /home/alice # 指定したディレクトリに移動（change directory）cd ..            # 一つ上のディレクトリに移動cd ~             # ホームディレクトリに移動```

```bashmkdir dataset            # dataset というディレクトリを作るmkdir -p data/train/img  # 途中のディレクトリごと一気に作るcp a.txt b.txt           # a.txt を b.txt という名前でコピーcp -r dir1 dir2          # ディレクトリ dir1 を丸ごとコピーmv old.txt new.txt       # 名前を変更（または移動）rm file.txt              # ファイルを削除rm -r dir                # ディレクトリを丸ごと削除```

## 2.3　ファイルの中身を見る

```bashcat log.txt        # ファイルの中身をすべて表示head -20 data.csv  # 先頭20行だけ表示tail -20 data.csv  # 末尾20行だけ表示tail -f train.log  # 追記されるログをリアルタイムに追う（学習の進捗確認に必須）less big.txt       # 大きなファイルをスクロールしながら閲覧（q で終了）wc -l data.csv     # 行数を数える（症例数の確認などに便利）```

## 2.6　プロセスとGPUの状態を見る

```bashtop              # CPUやメモリの使用状況をリアルタイム表示htop             # top の見やすい版（要インストール）nvidia-smi       # GPUの使用状況・VRAM・温度を表示watch -n 1 nvidia-smi  # 1秒ごとにGPU状態を更新表示ps aux | grep python   # 実行中のPythonプロセスを探すkill 12345       # プロセス番号12345を終了させる```

## 長時間の学習を安全に回す ― screen と nohup

```bash# screen を使う場合screen -S train        # "train" という名前の仮想端末を作るpython train.py        # この中で学習を開始# Ctrl+A → D で切り離す（学習は継続）。screen -r train で復帰# nohup を使う場合nohup python train.py > train.log 2>&1 &   # 閉じても継続、ログを記録tail -f train.log                          # 進捗をリアルタイムに追う```

## 手を動かす ― 演習

```bashcd ~                      # ホームに戻るmkdir -p medai/data       # 練習用フォルダを作るcd medaipwd                       # いまどこにいるか確認（/home/あなたの名前/medai と出る）echo "case001,骨折,陽性" > labels.csv   # ファイルを1行作るecho "case002,骨折,陰性" >> labels.csv  # 追記（>> は追記、> は上書き）cat labels.csv            # 中身を表示wc -l labels.csv          # 行数を数える（2 と出る）ls -l                     # 権限・サイズ付きで一覧```
