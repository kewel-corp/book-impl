## 高価な実験のアブレーション ― 安いプロキシと、症例ブートストラップ

```python
import numpy as np
def bootstrap_ci(per_case_dice, n=2000, alpha=0.05):
    d = np.asarray(per_case_dice)               # 症例ごとのDice
    boot = [np.mean(np.random.choice(d, len(d), replace=True)) for _ in range(n)]
    return np.mean(d), np.quantile(boot, [alpha/2, 1-alpha/2])
# 例：mean 0.49, 95%CI [0.41, 0.56]
# ※2つのモデルを比べるときは、それぞれのCIの重なりを見てはいけない。同じ症例で測る以上、
#   比較は「対応のある差」になる。患者ごとの差 d_i = A_i - B_i を作り、その差を再標本化して
#   差のCIを出す。各モデルのCIが重なっていても、差のCIが0をまたがないことは普通に起きる。
```
