## 4.2　Python本体のバージョン

```bash
python3 --version        # 例: Python 3.10.12
```

## 4.3　仮想環境（venv）― 標準の分け方

```bash
python3 -m venv .venv          # カレントに .venv という環境を作る
source .venv/bin/activate      # 有効化（Windowsは .venv\Scripts\activate）
# プロンプトの先頭に (.venv) が付けば成功。以後の pip はこの中だけに入る
pip install torch numpy pandas
deactivate                     # 環境から抜ける
```

## 4.4　conda / Miniconda ― データサイエンス向けの定番

```bash
conda create -n medai python=3.10     # medai という環境を Python3.10 で作る
conda activate medai                  # 有効化
conda install numpy pandas            # conda で入れる（pip との併用も可）
conda deactivate
```

## 4.5　pip・requirements.txt・uv ― パッケージを入れる／固定する

```bash
pip install torch==2.4.0 monai nibabel      # バージョンを指定して入れる
pip freeze > requirements.txt               # 今の環境の全ライブラリを記録
pip install -r requirements.txt             # 別の環境で同じ構成を再現
```

## requirements.txt の実例と、バージョン固定の作法

```text
# requirements.txt ― 3次元CTセグメンテーションの例
torch==2.4.0
torchvision==0.19.0
monai==1.3.2          # 医療画像向けディープラーニング
nibabel==5.2.1        # NIfTI(.nii.gz)の読み書き
pydicom==2.4.4        # DICOMの読み書き
numpy==1.26.4
scikit-image==0.24.0
matplotlib==3.9.2
```

## 4.7　GPU対応PyTorchのインストール ― つまずきの最頻ポイント

```bash

```

## 手を動かす ― 演習

```bash
python3 -m venv .venv           # 環境を作る
source .venv/bin/activate       # 有効化（プロンプト先頭に (.venv) が付く）
pip install numpy pandas nibabel   # 3つ入れてみる
pip freeze > requirements.txt   # 今の構成を記録
cat requirements.txt            # 中身を確認（3つ＋依存が並ぶ）
python -c "import nibabel; print(nibabel.__version__)"  # 動作確認
deactivate                      # 抜ける
```
