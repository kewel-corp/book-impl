## 4.3　仮想環境（venv）― 標準の分け方

```bashpython3 -m venv .venv          # カレントに .venv という環境を作るsource .venv/bin/activate      # 有効化（Windowsは .venv\Scripts\activate）# プロンプトの先頭に (.venv) が付けば成功。以後の pip はこの中だけに入るpip install torch numpy pandasdeactivate                     # 環境から抜ける```

## 4.4　conda / Miniconda ― データサイエンス向けの定番

```bashconda create -n medai python=3.10     # medai という環境を Python3.10 で作るconda activate medai                  # 有効化conda install numpy pandas            # conda で入れる（pip との併用も可）conda deactivate```

## 4.5　pip・requirements.txt・uv ― パッケージを入れる／固定する

```bashpip install torch==2.4.0 monai nibabel      # バージョンを指定して入れるpip freeze > requirements.txt               # 今の環境の全ライブラリを記録pip install -r requirements.txt             # 別の環境で同じ構成を再現```

## requirements.txt の実例と、バージョン固定の作法
